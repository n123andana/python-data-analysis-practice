```python
Q1 = df["Maths"].quantile(0.25) #first quantile ie 25th quantile
Q3 = df["Maths"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df["Maths"] < lower_bound) |
    (df["Maths"] > upper_bound)
]

outliers
```

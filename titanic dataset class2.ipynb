```python
#upper bound or lower bound
#upperbound=q3+1.5*IQR
#lowerbound=q1-1.5*IQR
#IQR=q3-q1
```

```python
Q1 = df["age"].quantile(0.25) #first quantile ie 25th quantile
Q3 = df["age"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
#QR method = a way of identifying unusually low or high values, not automatically proof that those values are wrong.
```

```python
#then find outliers ,outliers continuousl matre avullu , so age il and fare il aakum
outliers = df[
    (df["age"] < lower_bound) |
    (df["age"] > upper_bound)
]

outliers
```
